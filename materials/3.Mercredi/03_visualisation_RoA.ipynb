# Visualisation de données scientifiques réelles avec NumPy et Matplotlib

Dans ce notebook, nous allons visualiser un champ scientifique couvrant Madagascar et les régions voisines.

Nous utiliserons uniquement :

```python
import numpy as np
import matplotlib.pyplot as plt
```

Ces deux bibliothèques sont déjà disponibles avec Anaconda.

Les fichiers utilisés sont :

- `roa_lons_mdg.npy` : longitudes ;
- `roa_lats_mdg.npy` : latitudes ;
- `roa_202304101600.npy` : valeurs du champ RoA.

L'objectif est de construire progressivement une visualisation géographique claire à partir de vraies données.

## 1. Importer les bibliothèques

````python
import numpy as np
import matplotlib.pyplot as plt
````

## 2. Charger les données

Les trois fichiers sont des tableaux NumPy enregistrés au format `.npy`.

Utilisez `np.load()` pour les charger.

````python
lons = np.load("roa_lons_mdg.npy")
lats = np.load("roa_lats_mdg.npy")
roa = np.load("roa_202304101600.npy")
````

## 3. Vérifier les dimensions

Avant de tracer des données, il est important de vérifier que les tableaux sont compatibles.

````python
print("Longitudes :", lons.shape)
print("Latitudes  :", lats.shape)
print("RoA        :", roa.shape)
````

Les trois tableaux doivent avoir les mêmes dimensions.

Chaque valeur de `roa[i, j]` correspond donc aux coordonnées :

- longitude : `lons[i, j]`
- latitude : `lats[i, j]`

## 4. Examiner rapidement les valeurs

````python
print("Longitude min :", np.min(lons))
print("Longitude max :", np.max(lons))

print("Latitude min :", np.min(lats))
print("Latitude max :", np.max(lats))

print("RoA min :", np.min(roa))
print("RoA max :", np.max(roa))
````

## 5. Première visualisation avec `imshow`

`imshow` permet d'afficher directement un tableau 2D comme une image.

Cette première figure permet surtout de regarder la structure du champ, sans utiliser encore les coordonnées géographiques.

````python
plt.figure(figsize=(10, 8))

plt.imshow(roa, origin="lower")
plt.colorbar(label="Valeur RoA")

plt.xlabel("Indice x", fontsize=18)
plt.ylabel("Indice y", fontsize=18)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(False)

plt.show()
````

## 6. Visualisation géographique avec `pcolormesh`

Pour une carte scientifique, il est préférable d'utiliser les vraies longitudes et latitudes.

`pcolormesh` associe chaque valeur du tableau à ses coordonnées géographiques.

````python
plt.figure(figsize=(10, 10))

plt.pcolormesh(lons, lats, roa, shading="auto")

plt.colorbar(label="Valeur RoA")

plt.xlabel("Longitude", fontsize=18)
plt.ylabel("Latitude", fontsize=18)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(True, linewidth=0.1)

plt.show()
````

## 7. Améliorer la lisibilité

Une bonne visualisation scientifique doit permettre de comprendre rapidement :

- ce qui est représenté ;
- où se trouvent les données ;
- quelle quantité est associée aux couleurs.

On peut donc ajouter un titre et choisir une palette de couleurs adaptée.

````python
plt.figure(figsize=(10, 10))

plt.pcolormesh(
    lons,
    lats,
    roa,
    shading="auto",
    cmap="turbo"
)

cbar = plt.colorbar()
cbar.set_label("Valeur RoA", fontsize=16)
cbar.ax.tick_params(labelsize=14)

plt.title("Champ RoA — 10 avril 2023 à 16:00", fontsize=20)

plt.xlabel("Longitude", fontsize=18)
plt.ylabel("Latitude", fontsize=18)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(True, linewidth=0.1)

plt.show()
````

## 8. Se concentrer sur Madagascar

Les données couvrent une région plus large que Madagascar.

On peut utiliser `plt.xlim()` et `plt.ylim()` pour limiter la figure à une zone d'intérêt.

````python
plt.figure(figsize=(8, 10))

plt.pcolormesh(
    lons,
    lats,
    roa,
    shading="auto",
    cmap="turbo"
)

cbar = plt.colorbar()
cbar.set_label("Valeur RoA", fontsize=16)
cbar.ax.tick_params(labelsize=14)

plt.xlim(43, 51)
plt.ylim(-26, -11)

plt.title("Champ RoA autour de Madagascar", fontsize=20)

plt.xlabel("Longitude", fontsize=18)
plt.ylabel("Latitude", fontsize=18)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(True, linewidth=0.1)

plt.show()
````

## 9. Masquer les valeurs faibles

Dans certains cas, les petites valeurs rendent la figure difficile à lire.

NumPy permet de créer une copie du champ et de masquer les valeurs inférieures à un seuil.

Ici, les valeurs inférieures à `1` sont remplacées par `np.nan` uniquement pour la visualisation.

````python
roa_masque = roa.copy()
roa_masque[roa_masque < 1] = np.nan

plt.figure(figsize=(8, 10))

plt.pcolormesh(
    lons,
    lats,
    roa_masque,
    shading="auto",
    cmap="turbo"
)

cbar = plt.colorbar()
cbar.set_label("Valeur RoA", fontsize=16)
cbar.ax.tick_params(labelsize=14)

plt.xlim(43, 51)
plt.ylim(-26, -11)

plt.title("Valeurs RoA supérieures ou égales à 1", fontsize=20)

plt.xlabel("Longitude", fontsize=18)
plt.ylabel("Latitude", fontsize=18)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.grid(True, linewidth=0.1)

plt.show()
````

## 10. Localiser les valeurs les plus fortes

On peut utiliser un masque booléen pour sélectionner seulement les points dépassant un seuil.

Par exemple, cherchons les points où la valeur RoA est supérieure ou égale à `5`.

````python
fort = roa >= 5

print("Nombre de points avec RoA >= 5 :", np.sum(fort))
````

On peut ensuite afficher ces points avec `scatter` par-dessus le champ complet.

````python
plt.figure(figsize=(8, 10))

plt.pcolormesh(
    lons,
    lats,
    roa,
    shading="auto",
    cmap="Greys"
)

plt.scatter(
    lons[fort],
    lats[fort],
    s=8,
    label="RoA >= 5"
)

plt.xlim(43, 51)
plt.ylim(-26, -11)

plt.title("Localisation des valeurs RoA élevées", fontsize=20)

plt.xlabel("Longitude", fontsize=18)
plt.ylabel("Latitude", fontsize=18)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.legend(fontsize=14)

plt.grid(True, linewidth=0.1)

plt.show()
````
