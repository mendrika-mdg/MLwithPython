## 1. Importer Matplotlib


````python
import matplotlib.pyplot as plt
````

## 2. Courbe : suivre une évolution dans le temps

Une courbe est adaptée lorsque les observations ont un ordre naturel, par exemple des mesures successives.


````python
jours = [1, 2, 3, 4, 5, 6, 7]
niveau_riviere = [1.2, 1.4, 1.8, 2.6, 2.3, 1.9, 1.5]

plt.plot(jours, niveau_riviere, marker="o")

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.xlabel("Jour", fontsize=18)
plt.ylabel("Niveau de la rivière (m)", fontsize=18)
plt.grid(True, linewidth=0.1)
plt.show()
````

## 3. Mettre en évidence une valeur importante

On peut ajouter une annotation pour attirer l'attention sur un point particulier.


````python
plt.plot(jours, niveau_riviere, marker="o")
plt.annotate("Pic", xy=(4, 2.6), xytext=(4.5, 2.8), arrowprops={"arrowstyle": "->"})

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.xlabel("Jour", fontsize=18)
plt.ylabel("Niveau de la rivière (m)", fontsize=18)
plt.grid(True, linewidth=0.1)
plt.show()
````

## 4. Diagramme en barres : comparer des catégories

Un diagramme en barres permet de comparer des groupes distincts.


````python
disciplines = ["Biologie", "Physique", "Géologie", "Informatique"]
participants = [18, 12, 15, 25]

plt.bar(disciplines, participants)

plt.xticks(fontsize=16, rotation=20)
plt.yticks(fontsize=16)
plt.xlabel("Discipline", fontsize=18)
plt.ylabel("Nombre de participants", fontsize=18)
plt.grid(True, linewidth=0.1)
plt.show()
````

## 5. Histogramme : regarder une distribution

Un histogramme montre comment une variable numérique est répartie.


````python
tailles_plantes = [12, 14, 15, 15, 16, 17, 18, 18, 19, 20, 20, 21, 22, 23, 24, 25, 27, 29]

plt.hist(tailles_plantes, bins=6, edgecolor="black")

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.xlabel("Taille des plantes (cm)", fontsize=18)
plt.ylabel("Nombre de plantes", fontsize=18)
plt.grid(True, linewidth=0.1)
plt.show()
````

## 6. Nuage de points : relation entre deux variables

Chaque point correspond ici à un étudiant.


````python
heures_revision = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
notes = [8, 10, 9, 11, 13, 12, 14, 15, 16, 18]

plt.scatter(heures_revision, notes)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.xlabel("Heures de révision", fontsize=18)
plt.ylabel("Note / 20", fontsize=18)
plt.grid(True, linewidth=0.1)
plt.show()
````

## 7. Comparer plusieurs séries

On peut afficher plusieurs groupes sur une même figure lorsque les axes sont comparables.


````python
semaines = [1, 2, 3, 4, 5]
groupe_a = [12, 15, 18, 20, 23]
groupe_b = [10, 13, 17, 19, 21]

plt.plot(semaines, groupe_a, marker="o", label="Groupe A")
plt.plot(semaines, groupe_b, marker="s", label="Groupe B")

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.legend(fontsize=14)
plt.xlabel("Semaine", fontsize=18)
plt.ylabel("Score moyen", fontsize=18)
plt.grid(True, linewidth=0.1)
plt.show()
````

## 8. Modifier les limites des axes

Les limites peuvent aider à mieux voir une partie importante des données.


````python
plt.scatter(heures_revision, notes)
plt.xlim(0, 9)
plt.ylim(0, 20)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.xlabel("Heures de révision", fontsize=18)
plt.ylabel("Note / 20", fontsize=18)
plt.grid(True, linewidth=0.1)
plt.show()
````

## 9. Sauvegarder une figure

`savefig()` permet d'enregistrer une figure pour un rapport, une présentation ou un article.


````python
plt.bar(disciplines, participants)

plt.xticks(fontsize=16, rotation=20)
plt.yticks(fontsize=16)
plt.xlabel("Discipline", fontsize=18)
plt.ylabel("Nombre de participants", fontsize=18)
plt.grid(True, linewidth=0.1)

plt.savefig("participants_par_discipline.png", dpi=300, bbox_inches="tight")
plt.show()
````
