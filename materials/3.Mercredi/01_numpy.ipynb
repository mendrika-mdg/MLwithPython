NumPy est l'une des bibliothèques fondamentales de Python scientifique. Elle permet de travailler efficacement avec des collections de nombres sans écrire une boucle pour chaque opération.


## 1. Importer NumPy

Par convention, NumPy est importé sous le nom `np`.


````python
import numpy as np
````

## 2. Créer des tableaux

Un tableau NumPy, ou `array`, ressemble à une liste Python, mais il est conçu pour les calculs numériques.


````python
temperatures = np.array([24, 25, 27, 26, 23])
print(temperatures)
print(type(temperatures))
````

NumPy peut aussi créer automatiquement des suites de valeurs.


````python
heures = np.arange(0, 24, 3)
print(heures)

positions = np.linspace(0, 100, 6)
print(positions)
````

- `np.arange(debut, fin, pas)` crée une suite avec un pas régulier.
- `np.linspace(debut, fin, nombre)` crée un nombre fixé de valeurs régulièrement espacées.


## 3. Accéder aux valeurs et aux tranches

Comme pour une liste Python, l'indice commence à `0`.


````python
print(temperatures[0])
print(temperatures[-1])
print(temperatures[1:4])
````

## 4. La vectorisation : calculer sur tout un tableau

C'est l'un des principaux avantages de NumPy. Une opération appliquée à un tableau est appliquée à toutes ses valeurs.

### Exemple — conversion de températures

Avec NumPy, on peut convertir plusieurs températures en une seule instruction.


````python
celsius = np.array([0, 10, 20, 30, 40])
fahrenheit = celsius * 9 / 5 + 32

print("Celsius    :", celsius)
print("Fahrenheit :", fahrenheit)
````

Avec une liste Python classique, on utiliserait souvent une boucle. Avec NumPy, la formule s'applique directement à tout le tableau.


## 5. Calculs entre tableaux

Deux tableaux de même taille peuvent être combinés élément par élément.


````python
matin = np.array([20, 21, 22, 21, 19])
apres_midi = np.array([25, 26, 27, 26, 24])

ecart = apres_midi - matin
print(ecart)
````

Ici, chaque valeur du tableau `matin` est soustraite à la valeur correspondante de `apres_midi`.


## 6. Statistiques simples

NumPy fournit des fonctions pour résumer rapidement un ensemble de données.


````python
pluie = np.array([0, 12, 5, 0, 18, 7, 2])

print("Moyenne :", np.mean(pluie))
print("Minimum :", np.min(pluie))
print("Maximum :", np.max(pluie))
print("Total   :", np.sum(pluie))
print("Écart-type :", np.std(pluie))
````

L'écart-type donne une indication sur la dispersion des valeurs autour de la moyenne.


## 7. Filtrer des données avec des conditions

Une comparaison avec un tableau produit un tableau de valeurs booléennes (`True` ou `False`).


````python
temperatures = np.array([22, 27, 31, 29, 18, 34])

print(temperatures > 30)
````

On peut utiliser ce masque pour récupérer uniquement les valeurs qui satisfont la condition.


````python
chaudes = temperatures[temperatures > 30]
print(chaudes)
````

### Exemple — valeurs au-dessus et en dessous de la moyenne

Le *Python Workbook* propose un exercice où les valeurs sont séparées selon leur position par rapport à la moyenne. NumPy rend cette opération très directe.


````python
mesures = np.array([12, 15, 9, 18, 14, 11, 20])
moyenne = np.mean(mesures)

print("Moyenne :", moyenne)
print("Sous la moyenne :", mesures[mesures < moyenne])
print("Au-dessus de la moyenne :", mesures[mesures > moyenne])
````

## 8. Un exemple scientifique : chute libre

Le *Python Workbook* contient un problème de chute libre. Si un objet est lâché sans vitesse initiale, sa vitesse juste avant l'impact peut être estimée par


$v = \sqrt{2gh}$

où `g = 9.8 m/s²` et `h` est la hauteur.

Avec NumPy, on peut calculer la vitesse pour plusieurs hauteurs simultanément.


````python
hauteurs = np.array([1, 5, 10, 20, 50, 100])
g = 9.8

vitesses = np.sqrt(2 * g * hauteurs)

print("Hauteurs :", hauteurs)
print("Vitesses :", np.round(vitesses, 2))
````

## 9. Tableaux à deux dimensions

Un tableau peut contenir plusieurs lignes et plusieurs colonnes. Imaginons trois stations avec la température, l'humidité et la vitesse du vent.


````python
stations = np.array([
    [28, 72, 12],
    [31, 65, 18],
    [26, 80, 8]
])

print(stations)
print("Forme :", stations.shape)
````

Chaque ligne représente une station. Les colonnes représentent ici :

1. température ;
2. humidité ;
3. vitesse du vent.


````python
print("Première station :", stations[0])
print("Températures :", stations[:, 0])
print("Humidités :", stations[:, 1])
````

La notation `:` signifie « toutes les lignes ». Ainsi, `stations[:, 0]` sélectionne la première colonne pour toutes les stations.


## 10. Simulation avec des nombres aléatoires

Les simulations sont très utilisées en science. Le *Python Workbook* propose par exemple des simulations de pièces et de dés.

### Exemple — lancer un dé 20 fois


````python
rng = np.random.default_rng(42)
lancers = rng.integers(1, 7, size=20)

print(lancers)
print("Moyenne :", np.mean(lancers))
````

Le générateur est initialisé avec `42` pour obtenir les mêmes valeurs à chaque exécution du notebook. Cela est utile pour rendre une analyse reproductible.


### Exemple — deux dés

On peut simuler 1 000 lancers de deux dés et étudier la somme obtenue.


````python
de1 = rng.integers(1, 7, size=1000)
de2 = rng.integers(1, 7, size=1000)
sommes = de1 + de2

print("Premières sommes :", sommes[:20])
print("Somme moyenne :", np.mean(sommes))
print("Nombre de 7 :", np.sum(sommes == 7))
````
