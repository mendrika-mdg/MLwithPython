## 2. Variables et types

Une variable associe un nom à une valeur.


```python
age = 25
temperature = 28.5
ville = "Antananarivo"

print(type(age))
print(type(temperature))
print(type(ville))
```

## 3. Opérations

Python permet d'utiliser les opérateurs `+`, `-`, `*`, `/`, `//`, `%` et `**`.


```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

## 4. Chaînes de caractères

Les chaînes sont du texte. Elles peuvent être concaténées et indexées comme des séquences.


```python
prenom = "Mendrika"
print(prenom)
print(prenom[0])
print(prenom[-1])
print(len(prenom))
```

## 5. Listes et indexation

Une liste permet de stocker plusieurs valeurs dans un ordre défini. Le premier indice est `0`.


```python
temperatures = [24.5, 26.0, 28.3, 30.1, 27.8]

print(temperatures[0])
print(temperatures[-1])
print(temperatures[1:4])
print(len(temperatures))
```
