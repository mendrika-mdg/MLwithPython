# Lundi — Logique, boucles et fonctions

## Objectifs
- Écrire des conditions avec `if`, `elif` et `else`.
- Utiliser les opérateurs de comparaison et les booléens.
- Parcourir des valeurs avec une boucle `for`.
- Créer une fonction avec des paramètres et `return`.


## 1. Conditions


```python
temperature = 31

if temperature >= 30:
    print("Il fait chaud")
else:
    print("La température est inférieure à 30 °C")
```

## 2. Plusieurs cas avec `elif`


```python
note = 14

if note >= 16:
    print("Très bien")
elif note >= 12:
    print("Bien")
elif note >= 10:
    print("Passable")
else:
    print("À revoir")
```

## 3. Boucles `for`


```python
temperatures = [24, 27, 29, 31, 28]

for temperature in temperatures:
    print(temperature)
```

## 4. Fonctions

Une fonction permet de regrouper un calcul réutilisable. `return` renvoie le résultat.


```python
def carre(x):
    return x ** 2

print(carre(5))
```
