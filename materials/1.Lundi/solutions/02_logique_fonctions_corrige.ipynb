# Lundi — Logique, boucles et fonctions

## Objectifs
- Écrire des conditions avec `if`, `elif` et `else`.
- Utiliser les opérateurs de comparaison et les booléens.
- Parcourir des valeurs avec une boucle `for`.
- Créer une fonction avec des paramètres et `return`.


## 1. Conditions


```python
temperature = 31

if temperature >= 30:
    print("Il fait chaud")
else:
    print("La température est inférieure à 30 °C")
```

## 2. Plusieurs cas avec `elif`


```python
note = 14

if note >= 16:
    print("Très bien")
elif note >= 12:
    print("Bien")
elif note >= 10:
    print("Passable")
else:
    print("À revoir")
```

## 3. Boucles `for`


```python
temperatures = [24, 27, 29, 31, 28]

for temperature in temperatures:
    print(temperature)
```

## 4. Fonctions

Une fonction permet de regrouper un calcul réutilisable. `return` renvoie le résultat.


```python
def carre(x):
    return x ** 2

print(carre(5))
```

## Exercice 4 — Pair ou impair ?

Demandez un entier à l'utilisateur puis affichez s'il est **pair** ou **impair**.

Indice : le reste de la division par 2 s'obtient avec `%`.

**Notions :** `if`, `%`, comparaison.


```python
n = int(input("Entier : "))
if n % 2 == 0:
    print("Pair")
else:
    print("Impair")
```

## Exercice 5 — Voyelle ou consonne ?

Demandez une lettre à l'utilisateur.

- Si elle est `a`, `e`, `i`, `o` ou `u`, affichez **voyelle**.
- Si elle est `y`, affichez qu'elle peut être voyelle ou consonne.
- Sinon, affichez **consonne**.

**Notions :** chaînes, `if / elif / else`, opérateur `in`.


```python
lettre = input("Lettre : ").lower()
if lettre in ["a", "e", "i", "o", "u"]:
    print("Voyelle")
elif lettre == "y":
    print("Y peut être une voyelle ou une consonne")
else:
    print("Consonne")
```

## Exercice 6 — Table de conversion de température

Avec une boucle `for`, affichez une table de conversion de **0 °C à 100 °C**, par pas de **10 °C**.

Chaque ligne doit contenir la température en Celsius et son équivalent en Fahrenheit.

**Notions :** `for`, `range`, calculs.


```python
for celsius in range(0, 101, 10):
    fahrenheit = (9 / 5) * celsius + 32
    print(celsius, "°C =", fahrenheit, "°F")
```

## Exercice 7 — Fonction hypoténuse

Écrivez une fonction `hypotenuse(a, b)` qui reçoit les longueurs des deux côtés courts d'un triangle rectangle et renvoie la longueur de l'hypoténuse.

Utilisez le théorème de Pythagore :

$c = \sqrt{a^2 + b^2}$

Testez ensuite votre fonction avec plusieurs valeurs.

**Notions :** fonctions, paramètres, `return`, puissance.


```python
def hypotenuse(a, b):
    return (a ** 2 + b ** 2) ** 0.5

print(hypotenuse(3, 4))
print(hypotenuse(5, 12))
```
