## Exercice 8 — Calculer une moyenne

Écrivez un programme qui demande plusieurs valeurs à l'utilisateur.

La saisie s'arrête lorsque l'utilisateur entre `0`. Le `0` sert uniquement à arrêter la saisie et ne doit pas être inclus dans la moyenne.

Le programme doit ensuite afficher la moyenne des valeurs saisies.

Si la première valeur saisie est `0`, affichez un message indiquant qu'aucune moyenne ne peut être calculée.

**Niveau : intermédiaire**  
**Notions :** boucle, accumulateur, compteur, condition.


```python
total = 0.0
compteur = 0

valeur = float(input("Valeur (0 pour terminer) : "))
while valeur != 0:
    total += valeur
    compteur += 1
    valeur = float(input("Valeur (0 pour terminer) : "))

if compteur == 0:
    print("Aucune valeur saisie")
else:
    print("Moyenne :", total / compteur)
```

## Exercice 9 — Médiane de trois valeurs

Écrivez une fonction `mediane(a, b, c)` qui renvoie la valeur médiane de trois nombres.

La médiane est la valeur située au milieu lorsque les trois valeurs sont classées dans l'ordre croissant.

Essayez d'abord avec des conditions. Vous pouvez ensuite chercher une solution plus courte.

**Niveau : intermédiaire à avancé**  
**Notions :** fonction, paramètres, conditions, `return`.


```python
def mediane(a, b, c):
    if (a <= b <= c) or (c <= b <= a):
        return b
    if (b <= a <= c) or (c <= a <= b):
        return a
    return c

print(mediane(7, 2, 5))
print(mediane(1, 9, 3))
```

## Exercice 10 — Valeurs sous et au-dessus de la moyenne

Écrivez un programme qui demande à l'utilisateur de saisir plusieurs mesures. Une ligne vide termine la saisie.

Le programme doit :

1. stocker les mesures dans une liste ;
2. calculer leur moyenne ;
3. créer trois listes :
   - valeurs **inférieures** à la moyenne ;
   - valeurs **égales** à la moyenne ;
   - valeurs **supérieures** à la moyenne ;
4. afficher la moyenne et les trois groupes de valeurs.

Exemple de mesures : `12.4`, `15.7`, `11.2`, `18.3`, `14.9`.

**Niveau : avancé pour le lundi**  
**Notions :** listes, boucle, `input`, conditions, moyenne.


```python
mesures = []

while True:
    texte = input("Mesure (Entrée pour terminer) : ")
    if texte == "":
        break
    mesures.append(float(texte))

if len(mesures) == 0:
    print("Aucune mesure saisie")
else:
    moyenne = sum(mesures) / len(mesures)
    inferieures = []
    egales = []
    superieures = []

    for valeur in mesures:
        if valeur < moyenne:
            inferieures.append(valeur)
        elif valeur == moyenne:
            egales.append(valeur)
        else:
            superieures.append(valeur)

    print("Moyenne :", moyenne)
    print("Sous la moyenne :", inferieures)
    print("Égales à la moyenne :", egales)
    print("Au-dessus de la moyenne :", superieures)
```

# Mini-défi facultatif

Améliorez l'exercice 10 en ajoutant une fonction `moyenne(valeurs)` qui calcule et renvoie la moyenne d'une liste.

Puis remplacez le calcul direct de la moyenne par un appel à cette fonction.


```python
def moyenne(valeurs):
    return sum(valeurs) / len(valeurs)

mesures = [12.4, 15.7, 11.2, 18.3, 14.9]
print(moyenne(mesures))
```
