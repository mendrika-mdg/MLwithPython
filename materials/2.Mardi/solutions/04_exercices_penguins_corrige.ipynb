## 7. Premier histogramme

L'histogramme permet de voir comment les valeurs d'une variable sont réparties.

### Exercice 10

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(data["body_mass_g"].dropna(), bins=20)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)

plt.xlabel("Masse (g)", fontsize=18)
plt.ylabel("Nombre de manchots", fontsize=18)
plt.title("Distribution de la masse des manchots", fontsize=18)
plt.grid(True, linewidth=0.1)

plt.show()
```

La distribution présente plusieurs concentrations de valeurs. Cela peut notamment venir du fait que le jeu de données contient plusieurs espèces ayant des caractéristiques physiques différentes.

## 8. Un second histogramme

### Exercice 11

```python
plt.figure(figsize=(8, 5))
plt.hist(data["flipper_length_mm"].dropna(), bins=20)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)

plt.xlabel("Longueur des nageoires (mm)", fontsize=18)
plt.ylabel("Nombre de manchots", fontsize=18)
plt.title("Distribution de la longueur des nageoires", fontsize=18)
plt.grid(True, linewidth=0.1)

plt.show()
```

## 9. Relation entre deux variables

### Exercice 12

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    data["flipper_length_mm"],
    data["body_mass_g"]
)

plt.xticks(fontsize=16)
plt.yticks(fontsize=16)

plt.xlabel("Longueur des nageoires (mm)", fontsize=18)
plt.ylabel("Masse (g)", fontsize=18)
plt.grid(True, linewidth=0.1)

plt.show()
```

Le nuage de points montre une relation positive : les manchots avec des nageoires plus longues ont généralement une masse plus élevée.
