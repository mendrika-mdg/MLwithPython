## Exercice 1 — Anagrammes


```python
def sont_anagrammes(mot1, mot2):
    mot1 = mot1.lower()
    mot2 = mot2.lower()
    return sorted(mot1) == sorted(mot2)

print(sont_anagrammes("evil", "live"))
print(sont_anagrammes("python", "typhon"))
```

## Exercice 2 — Score d’un mot


```python
points = {
    "a": 1, "e": 1, "i": 1, "l": 1, "n": 1, "o": 1, "r": 1, "s": 1, "t": 1, "u": 1,
    "d": 2, "g": 2,
    "b": 3, "c": 3, "m": 3, "p": 3,
    "f": 4, "h": 4, "v": 4, "w": 4, "y": 4,
    "k": 5,
    "j": 8, "x": 8,
    "q": 10, "z": 10
}

def score_mot(mot):
    total = 0
    for lettre in mot.lower():
        total += points.get(lettre, 0)
    return total

print(score_mot("python"))
```

## Exercice 3 — Chiffrement de César


```python
def cesar(message, decalage):
    resultat = ""

    for caractere in message:
        if caractere.isalpha() and caractere.isascii():
            base = ord("A") if caractere.isupper() else ord("a")
            resultat += chr((ord(caractere) - base + decalage) % 26 + base)
        else:
            resultat += caractere

    return resultat

message = "Python pour la science!"
chiffre = cesar(message, 4)

print(chiffre)
print(cesar(chiffre, -4))
```

## Exercice 4 — Défi avancé : supprimer les valeurs extrêmes


```python
def supprimer_extremes(valeurs, n):
    if n < 0:
        raise ValueError("n doit être positif ou nul.")

    if 2 * n >= len(valeurs):
        raise ValueError("La liste ne contient pas assez de valeurs.")

    triees = sorted(valeurs)
    return triees[n:len(triees) - n]

mesures = [14.2, 13.8, 50.0, 14.5, 14.0, -10.0, 13.9, 14.1]

print(supprimer_extremes(mesures, 1))
print(mesures)
```
