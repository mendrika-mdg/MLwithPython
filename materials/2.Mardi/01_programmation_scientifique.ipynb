## 1. Dictionnaires
Un dictionnaire associe des clés à des valeurs. Il représente naturellement les attributs d’un échantillon. Une clé absente provoque une erreur ; `get` permet de fournir une valeur par défaut.

````python
echantillon = {"id": "A01", "masse_g": 12.5, "valide": True}

print(echantillon["masse_g"])
print(echantillon.get("site", "non renseigné"))

for cle, valeur in echantillon.items():
    print(cle, valeur)
````

## 2. Tuples et décomposition
Un tuple est une collection ordonnée dont les éléments ne sont pas réaffectables. La décomposition donne un nom à chaque composante.

````python
coordonnees = (-18.9, 47.5)
latitude, longitude = coordonnees

print(latitude, longitude)
````

## 3. Indices et paires
`enumerate` fournit l’indice et la valeur. `zip` parcourt plusieurs collections ensemble ; par défaut, il s’arrête à la plus courte. Vérifiez donc leurs longueurs.

````python
noms = ["A", "B", "C"]
valeurs = [12.5, 14.2, 11.8]

assert len(noms) == len(valeurs)

for indice, nom in enumerate(noms):
    print(indice, nom)
    
for nom, valeur in zip(noms, valeurs):
    print(nom, valeur)
````

## 4. Compréhensions de listes
Une compréhension compacte une transformation ou une sélection simple. Une boucle classique reste préférable lorsque le traitement devient difficile à lire.

````python
valeurs_doubles = [2 * valeur for valeur in valeurs]
valeurs_selectionnees = [valeur for valeur in valeurs if valeur > 12]

print(valeurs_doubles, valeurs_selectionnees)
````

## 5. Boucle while
`while` répète tant que la condition est vraie. Il faut garantir une condition d’arrêt. Une limite d’itérations protège les calculs itératifs.

````python
erreur = 1.0
iteration = 0

while erreur > 0.01 and iteration < 20:
    erreur = erreur / 2
    iteration += 1
    
print(iteration, erreur)
````

## 6. Contrats d’une fonction
Précisez les entrées, le résultat et les cas invalides. Une liste vide ne permet pas de calculer une moyenne : nous signalons explicitement ce problème.

````python
def resumer(valeurs):
    
    if len(valeurs) == 0:
        raise ValueError("La liste de mesures est vide.")
        
    return {"effectif": len(valeurs), 
            "moyenne": sum(valeurs) / len(valeurs), 
            "minimum": min(valeurs), 
            "maximum": max(valeurs)
           }

print(resumer(valeurs))
````
